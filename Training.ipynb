```python
import pandas as pd
df = pd.DataFrame({
    'gender': ['male', 'female', 'female', 'male']
})
mapping = {'male': 'Мужчина', 'female': 'Женщина'}
df['gender_null'] = df.gender.map(mapping)
df
```

```python
df = pd.DataFrame({
    'gender': ['male', 'female', 'female', 'male']
})
df['gender_upper'] = df['gender'].map(str.upper)
df
```

```python
import pandas as pd
df = pd.DataFrame({
    'score': [10, 25, 40]
})


df['score_sq'] = df['score'].apply(lambda x: x ** 2)
df
```

```python
import pandas as pd
df = pd.DataFrame({
    'price': [100, 250, 400],
    'discount': [10, 0, 50]
})

def calc(row):
    return row['price'] - row['discount']

df['final_price'] = df.apply(calc, axis=1)
df
```
